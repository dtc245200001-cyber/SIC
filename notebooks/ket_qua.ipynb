# Ket qua - Phat hien binh luan doc hai tieng Viet (ViHSD)

Notebook nay CHI doc file trong `outputs/`. Neu thieu file, hay chay script tuong ung truoc (`scripts/01` -> `05`).

```python
import os
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from IPython.display import Image, display

OUT = "../outputs"

def ton_tai(ten):
    p = os.path.join(OUT, ten)
    if not os.path.exists(p):
        print("Chua co ket qua (%s), hay chay script truoc" % ten)
        return None
    return p
```

## 1. Bang so sanh mo hinh (tap test)

```python
p = ton_tai("bang_so_sanh_mo_hinh.csv")
if p:
    bang = pd.read_csv(p)
    display(bang)
    if {"mo_hinh", "f1_macro"}.issubset(bang.columns):
        plt.figure(figsize=(6, 4))
        sns.barplot(data=bang, x="mo_hinh", y="f1_macro")
        plt.title("F1-macro tren tap test")
        plt.tight_layout()
        plt.show()
```

## 2. Confusion matrix (tap test)

```python
p = ton_tai("confusion_matrix_test.png")
if p:
    display(Image(filename=p))
```

## 3. Phan tich loi (tap val)

```python
p = ton_tai("phan_tich_loi.csv")
if p:
    loi = pd.read_csv(p)
    print("Tong so cau doan sai: %d" % len(loi))
    if "mo_hinh" in loi.columns:
        print(loi.groupby("mo_hinh").size())
    display(loi.head(20))
q = ton_tai("ma_tran_cap_nham_val.csv")
if q:
    print("Ma tran cap nham:")
    display(pd.read_csv(q))
```
